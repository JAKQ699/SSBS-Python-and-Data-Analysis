## 2.1 Basic Language Semantics

### 2.1.1 Indentation, not braces
Python uses whitespace (tabs or spaces) to structure code instead of using braces as in many other languages like R, C++, Java, and Perl.

### 2.1.2 Everything is an object
Each object has an associated type (e.g. integer, string, or function) and internal data, which makes the language very flexible.

### 2.1.3 Comments
Make sure you write good comments when coding as they will make your code much more legible.

### 2.1.4 Variables and argument passing
When assigning a variable in Python, you are creating a reference to the object, a process called "aliasing".

Sometimes, aliasing can cause some troubles, for example when you are making copies.

### 2.1.5 Types, attributes, and methods
Python is sometimes referred to as a "strong type" language, as objects in Python have specific types (though variables can be flexibly assigned with different types) that typically do not communicate with one another directly, which we'll look at in more detail in **2.2**.

Besides, objects in Python also typically have attributes (other objects stored "inside" the object) and methods (functions associated with the object).

### 2.1.6 Imports
We need to import variables and functions from another file in the same directory in order to use them.

### 2.1.7 Binary operators and comparisons
Most binary operations are similar to other programming languages.

## 2.3 Data Types and Structures

### 2.3.1 Data types (scalars)
Python has a small set of built-in types for handling numerical data, strings, Booleans, and dates and time (dates and times will be discussed separately).

#### Numeric types: int and float
You need to be careful with float operations sometimes. Consider the following famous example: 0.1 + 0.2 ?= 0.3

#### String
Strings can sometimes behave like lists, like len(), indexing, and concatenation.

#### Formatted string
Convenient for replacing variables in the string.

#### Booleans
Combined with and, or, not operations.

#### None
Python null value type, coupled with "is" (we have different null types for NumPy and pandas).

#### Type casting
Cast between str, bool, int, and float types (need to be careful sometimes, and consider math.floor and math.ceil)

### 2.3.2 Data structures (vectors)
Tuple, list, and dictionary are some of the most frequently used data structures.

#### Tuple
Tuples are **immutable data type** which, once assigned, cannot be changed (unless the elements in the tuple are mutable).

#### List
Lists are **mutable data type** which are variable length and their contents an be modified in place.

#### Set
Sets are unordered collections of unique elements. They are also **mutable data type**.

#### Dictionary
Dictionaries store **key-value pairs**, and keys need to be immutable (and hashable).

#### Nested list, set, and dictionary comprehensions
Instead of using a lot of loops, you may choose the shortcut for creating lists, sets, or dictionaries.
