# Unit 3. NumPy Basics

NumPy (Numerical Python) is one of the core packages for numerical computing in Python. Pandas, Matplotlib, Statmodels and many other Scientific libraries rely on NumPy.

NumPy major contributions are:

* Efficient numeric computation with C primitives
* Efficient collections with vectorized operations
* An integrated and natural Linear Algebra API
* A C API for connecting NumPy with libraries written in C, C++, or FORTRAN.

Let's develop on efficiency. In Python, **everything is an object**, which means that even simple ints are also objects, with all the required machinery to make object work. We call them "Boxed Ints". In contrast, NumPy uses primitive numeric types (floats, ints) which makes storing and computation efficient.

## 3.1 Creating ndarrays
An ndarray is a generic multidimensional container for homogeneous data, where each array has a **shape** and **dtype**. With functions like np.zeros, you are able to create uni- or multi-dimenstional numerical arrays efficiently.

```python
arr_1d = np.array([1, 2, 3, 4])
print(arr_1d)
print(arr_1d.shape)
print(arr_1d.dtype) # default of integers: int64
arr_1d.dtype == int

arr_1d_2 = np.array([1, 2.5, 3, 4]) # default of float: float64
print(arr_1d_2.dtype)

arr_1d_3 = np.array([1, 2.5, "3", 4])
print(arr_1d_3.dtype)

matrix = [[1, 2, 3], [4, 5, 6], [7, 8, 9]]
arr_2d = np.array(matrix, dtype="int8") # np.int8
print(arr_2d)
print(arr_2d.shape) # (row, col)
print(arr_2d.dtype)
print(arr_2d.ndim) # number of dimensions

arr_range_py = list(range(10))
arr_range_np = np.arange(10)
print(arr_range_np)
arr_range_np_2 = np.arange(0, 100, 10) # start, end, step
print(arr_range_np_2)
# cannot do range(0, 1, 0.1)
arr_range_np_3 = np.arange(0, 1, 0.1) # start, end, step
print(arr_range_np_3)
arr_range_np_3[3] == 0.3 # Careful about float precision

arr_range_np.reshape((5, 2))
np.arange(64).reshape(8, 8)

np.zeros(9).reshape(3, 3)
np.ones(32).reshape(4, 8)
np.identity(4)
np.eye(5, 4)
```

## 3.2 Data Types for ndarrays
Data types in ndarrays, like int 32 or int 64, is not the same as, but compatible with, the default data types in Python (as they provide mapping directly to underlying memory representation). You are able to cast between types with **.astype()**.

```python
# Integers: int8, int16, int32, int64
# Floats: float8, float16, float32, float64
# Unsigned integers: uint8, uint16, uint32, uint64
# Booleans: np.True_, np.False_
# Unicode strings (UCS-4 encoding): np.str_ / "U"
# Byte strings (ASCII encoding): np.bytes_ / "S"
arr1 = np.array([1, 2, 3, 4, 5])
arr1.astype(np.int8)

# uint in numpy
arr2 = np.array([-1, 2, 3, -4, 5])
arr2.astype(np.uint8) # uint8: 0 to 255

arr3 = np.array([256, 2, 3, 258, 5])
arr3.astype(np.uint8)

arr_str = np.array(["Alice", "Bob", "Eric", "Linda"])
arr_str.astype(np.bytes_) # strings
print(arr_str[0])
arr1.astype(np.bytes_)
# arr_str.astype(np.int64)

arr_bool = np.array([True, False, True, False, True])
arr_bool.astype(np.bool_)
arr_bool[1]
arr_bool.astype(np.int8)

# np.nan is equivalence of python None
arr_NA = np.array([1, 2, None, 4, 5])
print(arr_NA[2])
arr_nan = np.array([1, 2, np.nan, 4, 5]) # Recommended in numpy
print(arr_nan[2])
arr_NA[2] == arr_nan[2] # But not exactly the same
```

## 3.3 Basic Indexing and Slicing
Indexing of ndarray works similar to indexing of multidimensional lists in Python, and is more flexible.

```python
# 1d numpy array
# np_1d[index]
# np_1d[start: end]
# np_1d[start: end: step]
str_array = np.array(list("abcdefghijklmn"))
print(str_array[5])
print(str_array[3:16:3])
print(str_array[::-1])
# multi-indexing with some order: np_array[list]
print(str_array[[1, 3, 5, 9]])
print(str_array[[3, 9, 5, 1]])

# 2d numpy array
# np_2d[row][col]
# np_2d[row, col]
# np_2d[index, progression] # Mix index, progression, list
array_2d = np.arange(24).reshape(6, 4)
print(array_2d)
print(array_2d[3, 2])
# all the elements of a certain row/col
print(array_2d[2, :])
print(array_2d[:, 1])
print(array_2d[:, [1, 2]])

# Create an n x n checkerboard: alternating 1s and 0s
def checkerboard(n):
    board = np.zeros((n, n))
    board[::2, ::2] = 1
    board[1::2, 1::2] = 1
    return board

checkerboard(5)
```

## 3.4 Boolean Indexing
What's cool is that you can add conditional statements to indexing of ndarrays--think of it as applying a mask. So you can write something like **data[condition1 and condition2]**, which can be mixed with basic indexing.

```python
data = [1, 2, 4, None, 6, None, 7]
data_np = np.array([1, 2, 4, None, 6, None, 7])
# Filter None values
data_filtered = [x for x in data if x != None]
data_np_filtered = data_np[data_np != None] # Broadcasting: element-wise operation
print(data_filtered)
print(data_np_filtered)

data_2d = np.arange(48).reshape(6, 8)
data_2d[2:4, [1, 3, 5]] = -1
print(data_2d)
data_2d[data_2d < 10]

students = np.array(["Alice", "Bob", "Eric", "Linda", "Sam"])
grades = np.array([90, 50, 80, None, 60])
grades_2 = np.array([90, np.nan, 80, 50, 60])
# Students who fail (grade < 60) or didn't attend (None)
print(grades is None) # This is not broadcasting
print(grades == None) # Boolean list
print(students[grades == None]) # apply this boolean list as a mask
print(students[~(grades == None)]) # Replace not with ~
print(students[grades_2 < 60])
print(students[(np.isnan(grades_2)) | (grades_2 < 60)]) # and, or replaced with &, |, be careful there's no short circuit!
# Cannot say: grades_2 == np.nan

# Strongly recommended: replace None with np.nan
```

## 3.5 ndarray Operations
Instead of writing excessive loops, NumPy allows you to perform quick element-wise or array-wise operations, a process often called **"broadcasting"**

### 3.5.1 Element-wise ndarray operations
Element-wise ndarray operations include simple arithmetics (+, *), unary universal-functions (ufuncs) like np.exp and np.isnan, binary ufuncs like np.add and np.fmax, and conditional logics like np.where.

```python
lst_py = [1, 2, 3, 4, 5]
lst_big = list(range(1000))
lst_np = np.array(lst_py)
lst_big_np = np.array(lst_big)
lst_py * 4 # Makes n copies of the list
lst_2d = np.arange(16).reshape(4, 4)
lst_nan = np.array([1, 2, np.nan, 4, 5])
lst_1 = np.array([1, 6, 7, 9, 5])
lst_2 = np.array([6, 2, 8, 4, 10])

# Unary operations
print([x * 4 for x in lst_py])
print(lst_np * 4)

%timeit -n 100 [x ** 2 for x in lst_big]
%timeit -n 100 lst_big_np ** 2

print(np.sqrt(lst_np))
print(np.exp(lst_np))
print(np.log(lst_np)) # log2 log10
print(np.round(np.sqrt(lst_np), 2)) # np.round(array, decimals)
print(np.round(np.sqrt(lst_2d), 2)) # 2-layer broadcasting: matrix -> list -> element
print(np.isnan(lst_nan))

# Binary operations
print(lst_1 + lst_2)
print(np.add(lst_1, lst_2))
print(lst_1 * lst_2)
print(np.multiply(lst_1, lst_2))
print(np.subtract(lst_1, lst_2)) # lst_1 - lst_2
print(np.subtract(lst_2, lst_1)) # lst_2 - lst_1
print(np.maximum(lst_1, lst_2)) # Compare lst_1 and lst_2 element-wise
print(np.minimum(lst_1, lst_2))
print(np.maximum(lst_nan, lst_2)) # will not ignore the nan value
print(np.fmax(lst_nan, lst_2)) # Will ignore the nan value
print(np.fmin(lst_nan, lst_2))

# Conditional operations
# np.where(condtion, list1, list2): [x for x in list1 if condition else list2]
def removeNan(np_lst): # replace each np.nan with -1
    return np.where(np.isnan(np_lst), -1, np_lst)

def findMax(list1, list2): # Does the same thing as np.maximum
    return np.where(list1 >= list2, list1, list2)

def findMin(list1, list2): # Does the same thing as np.minimum
    return np.where(list1 >= list2, list2, list1)

def findMax_nan(list1, list2): # Does the same thing as np.fmax
    return np.where((~(np.isnan(list1))) & (list1 >= list2), list1, list2)

print(removeNan(lst_nan))
print(findMax(lst_1, lst_2))
print(findMin(lst_1, lst_2))
print(findMax_nan(lst_nan, lst_1))
```

### 3.5.2 Array-wise ndarray operations
Array-wise operations are useful in mathematical and statistical methods, and allows aggregation along a single axis (0 or 1). We'll explore a few commonly used ones:

* Statistical methods: np.mean(), np.sort(), np.unique() (beware of np.nan values)
* Boolean arrays: [condition].sum(), [condition].any(), [condition].all()

```python
# Grades: row is student, col is exam
grades = np.array([[57, 68, 72, 53, 79],
                [84, 77, 56, 66, 67],
                [92, 81, np.nan, 85, 91],
                [89, 79, 75, 76, 76],
                [93, 100, 88, np.nan, 93],
                [57, 90, 89, 89, 77]])
students = np.array(["Bob", "Amy", "Lucy", "Amanda", "Cindy", "Eric"])
exams = np.array(["Exam1", "Exam2", "Midterm", "Exam3", "Final"])
# Task1: find students indexed 2, 3 with exam grades indexed 1-3
print(grades[[2, 3], 1:4])
# Task2: find the midterm and final grade for Lucy
print(grades[(students == "Lucy"), ((exams == "Midterm") | (exams == "Final"))])

# Task3: calculate the average grade for each exam across the students
# Task4: calculate the average grade for each student across the 5 exams
np.nanmean(grades) # Element-wise operation, nan ignores the nan values
print(np.nanmean(grades, axis=0)) # task3
print(np.nanmean(grades, axis=1)) # task4

# Task5: sort the students by grade for each exam
print(np.sort(grades, axis=0)) # ascending order by default, with nan values at the bottem
print(np.sort(grades, axis=0)[::-1])

# Task6: Find out whether each exam has absent students
# Task7: Find out whether each student is absent for some exam
print(np.isnan(grades).any(axis=0)) # task6
print(np.isnan(grades).any(axis=1)) # task7

# Task8: Find the students with at least 2 absences or fails
counts = ((grades == np.nan) | (grades < 60)).sum(axis=1)
print(students[counts >= 2])
```

## 3.6 Pseurandom Number Generation
Random numbers are not truly random but instead are generated by a configurable random number generator in NumPy (which is typically faster than the Python random). Commonly used operations include:

* np.random.standard_normal(size)
* np.random.randint(low, high, size)
* np.random.default_rng(seed) (to configure a "fixed" rng)

```python
print(np.random.standard_normal(10)) # standard normal distribution
normal_m = np.random.standard_normal((6, 4))
print(normal_m)

# Replace every negative of this normal matrix with 0
print(np.where(normal_m < 0, 0, normal_m))

# Create a 10 x 10 random matrix with values between 1 and 100
print(np.random.randint(1, 101, size=(10, 10)))

# Create a 10 x 10 random matrix with values between 0 and 1
np.random.uniform(size=(10, 10))

rng = np.random.default_rng(seed=12345) # random number generator, fix seed will fix the number generated
rng.integers(100, size=10)

# Random walk: create a walk of nsteps steps, each step can take -1 and 1; walk is cumulative sum of steps
import matplotlib.pyplot as plt

def randomWalk(nsteps):
    steps = rng.integers(2, size=nsteps)
    steps_true = np.where(steps == 0, -1, 1)
    walk = np.cumsum(steps_true) # cumulative sum of steps
    return walk

walk1 = randomWalk(100)
plt.plot(walk1)
print(np.max(walk1)) # Differentiate this from np.maximum
print(np.min(walk1))
print(np.argmax(np.abs(walk1))) # the index of the maximium of absolute value

def randomWalk_2d(nwalks, nsteps): # a series of walks
    steps = rng.integers(2, size=(nwalks, nsteps))
    steps_true = np.where(steps == 0, -1, 1)
    walk = np.cumsum(steps_true, axis=1) # cumulative sum of steps, aggregating along the column
    return walk

randomWalk_2d(5, 100)
```

## 3.7 Linear Algebra
NumPy supports linear algebra operations, like matrix multiplication, transposition, determinants, etc. Feel free to explore matrix operations on your own~

```python
matrix_1 = np.arange(9).reshape(3, 3)
matrix_2 = np.arange(9).reshape(3, 3)
matrix_2[0, 0] = 1
matrix_2[2, 2] = 6

# Addition
print(matrix_1 + matrix_2)

# Multiplication
print(matrix_1 * matrix_2) # element-wise multiplication
print(matrix_1 @ matrix_2) # matrix multiplication
print(np.dot(matrix_1, matrix_2)) # matrix multiplication
print(np.dot(matrix_2, matrix_1))

# Transposition
print(matrix_1.T)
print(matrix_1.swapaxes(0, 1))

# np.linalg
np.linalg.det(matrix_2)
np.linalg.inv(matrix_2)
```
