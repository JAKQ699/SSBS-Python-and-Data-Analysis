### 2.1.2 Everything is an object
Each object has an associated type (e.g. integer, string, or function) and internal data, which makes the language very flexible.

```python
# Functions are object
def sqr(num): # no need to define input/output types
    return num ** 2

def applyTwice(num, f): # f is a function
    return f(f(num))

def applyNTimes(num, f, n=1): # n: number of times to apply f, with default value 1
    for i in range(n):
        num = f(num)
    return num

print(applyTwice(3, sqr))
print(applyNTimes(2, sqr, 4))
```

## 2.2 Control Flow

### 2.2.1 if, elif, and else

```python
# if (condition1):
#    do something 1
# elif (condition2):
#    do something 2
# elif (condition3):
#    do something 3
# elif (condition4):
#    do something 4
# else:
#    do something 

def getGrade(score):
    if (score >= 90):
        return "A"
    elif (score >= 80):
        return "B"
    elif (score >= 70):
        return "C"
    elif (score >= 60):
        return "D"
    else:
        return "F"

getGrade(65)
```

```python
# Short circuit
# if (condition1) and (condition2)
# if False and (condition2: Error)
# if True or (condition2: Error)
def getGrade2(score):
    if (type(score) in [int, float]) and (score >= 90):
        return "Good"
    return "Bad"

getGrade2("90")
```

```python
# continue, break, pass, return
# continue: skip this iteration of the loop
# pass: do nothing
# break: exit the entire loop
# return: exit the function

sequence = [1, 2, None, 4, None, 5, "Stop", "foo", 6]

def sum_1(seq):
    s = 0
    for i in range(len(seq)):
        if seq[i] is None:
            continue # skip this iteration
        elif seq[i] == "Stop":
            break # exit the loop
        else:
            s += seq[i]
        print(f"Iteration {i}: current sum is {s}")
    return s

def sum_2(seq):
    s = 0
    for i in range(len(seq)):
        if seq[i] is None:
            pass # do nothing
        elif seq[i] == "Stop":
            break # exit the loop
        else:
            s += seq[i]
        print(f"Iteration {i}: current sum is {s}")
    return s

sum_1(sequence)
sum_2(sequence)
```

### 2.2.3 try and except

```python
# try:
#   do something 1 # Raises an error, go to except
# except:
#   do something 2

# try:
#   do something 1 # Raises an error, go to except
# except:
#   do something 2
# else:
#   do something 3 # No error
# finally:
#   do something 4 # Regardless of error

def sqr_plus(num): 
    try:
        return num ** 2
    except TypeError:
        return "Invalid input"

def sqr_ultra(num): 
    try:
        print(num ** 2)
    except TypeError:
        print("Fail")
    else:
        print("Succeed")
    finally:
        print("Done")
        
sqr_plus("2")
sqr_ultra(4)
```

```python
# Indexing and slicing
m = [[1, 2, 3],
     [4, 5, 6],
     [7, 8, 9]]
def flatten(matrix): # convert a 2D matrix to a 1D list
    lst = []
    for row in matrix:
        for item in row:
            lst.append(item)
    return lst

flatten(m)

# lst[index]
# lst[start: end]
# lst[start: end: step]
# Revert a list
print(m[::-1])
list(reversed(m))
```

```python
# List operations
# append(element): add element to list at the last position
# insert(index, element): add element to index-position
# extend(lst): add a list of elements to another list
# pop(index=last): remove the index-position element
# remove(element): remove the element by name

guests = ["Amy", "Bob", "Cindy", "Derek", "Eric"] # maximum occupancy
queue = ["Bob", "Peter", "Sam"]

def empty_Queue(guest_lst, queue_lst): # remove everyone from queue_lst and add to guest_lst
    '''
    while (queue_lst != []):
        guest_lst.append(queue_lst.pop(0))
    '''
    guest_lst.extend(queue_lst)
    queue_lst = []

def deQueue(guest_lst, queue_lst, n=1):
    num = min(n, len(queue_lst))
    for _ in range(num):
        guest_lst.append(queue_lst.pop(0))

def leave(guest_lst, max_occupancy):
    num = len(guest_lst) - max_occupancy
    if num > 0:
        for _ in range(num):
            guest_lst.pop(0)

deQueue(guests, queue, 2)
leave(guests, 6)
print(guests)
print(queue)
```

#### Dictionary
Dictionaries store **key-value pairs**, and keys need to be immutable (and hashable).

```python
empty_dict = dict() # {}
# dict = {key1: value1, key2: value2}
# Two attributes: dict.keys(), dict.values()
dict_1 = {"a": 1, 1: "a", ("a", 1): [1, 2, "foo"]} # Mutable objects cannot be keys
print(dict_1.keys())
print(dict_1.values())
print(dict_1["a"])

category_1 = ["fruit", "veggie", "protein", "carbo"]
object_1 = ["apple", "cabbage", "chicken", "rice"]
# dict = {category: object}
food_dict = dict(zip(category_1, object_1))
print(food_dict)

category_2 = ["veggie", "fruit", "veggie", "protein", "protein", "carbo", "veggie", "fruit"]
object_2 = ["lettuce", "apple", "cabbage", "chicken", "beef", "rice", "eggplant", "banana"]
# dict = {category: [object1, object2, ...]}
def makeDict_1(key_lst, value_lst): # Make sure the two lists have the same length
    new_dict = {}
    for i in range(len(key_lst)):
        if key_lst[i] in new_dict.keys():
            new_dict[key_lst[i]].append(value_lst[i])
        else:
            # create a new key with a new list of objects
            new_dict[key_lst[i]] = [value_lst[i]]
    return new_dict

food_1 = makeDict_1(category_2, object_2)

def makeDict_2(key_lst, value_lst):
    new_dict = defaultdict(list) # takes in a type that becomes the type of value
    for i in range(len(key_lst)):
        new_dict[key_lst[i]].append(value_lst[i])
    return new_dict

print(makeDict_2(category_2, object_2))

# Delete keys
del food_1["carbo"]
food_1.pop("fruit")
food_1
```

### 2.4.2 Functions are objects
Functions are objects that can be passed on to other functions themselves!

```python
students = ["Alice", "Amy", "Tom", "Cindy", "Eric", "Bob", "Cynthia", "Amanda"]
# itertools.groupby(data, key) is going to return a list of (group key, group value)
def groupByAlphabet(lst):
    # Do sorting first
    lst_sorted = sorted(lst)
    for key, group in itertools.groupby(lst_sorted, key=lambda x: x[0]): # first letter of the string
        print(f"Group {key}: {list(group)}")

def groupByLength(lst):
    # Do sorting first
    lst_sorted = sorted(lst, key=len)
    for key, group in itertools.groupby(lst_sorted, key=len): # first letter of the string
        print(f"Length {key}: {list(group)}")

groupByAlphabet(students)
print("\n")
groupByLength(students)
```

### 2.4.3 Anonymous (Lambda) functions
Shortcut for writing short functions, without explicitly declaring the function name.

```python
# lambda inputs: outputs
'''
def applyTwice(num, f): # f is a function
    return f(f(num))
'''
applyTwice(4, lambda x: x ** 2)

# lambda x, y: f(x, y)
sum_lambda = lambda x, y: x + y # sum

# lambda input: output1 if condition else output2
'''
def getGrade2(score):
    if (type(score) in [int, float]) and (score >= 90):
        return "Good"
    return "Bad"
'''

getGrade_lambda = lambda score: "Good" if (type(score) in [int, float]) and (score >= 90) else "Bad"
print(getGrade_lambda("f00"))

# lambda input_list: output for input in input_list
'''
def sqr_2(lst):
    return [round(x ** 2) for x in lst if type(x) in [int, float]]
'''

sqr_lambda = lambda lst: [round(x ** 2) for x in lst if type(x) in [int, float]]
print(sqr_lambda([1, 2, 3.2, "foo", None, 4, None, "SSBS"]))
```
