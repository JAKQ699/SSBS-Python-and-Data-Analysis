For this course, we'll be mainly focusing on **Python** and its various libraries/packages for data analysis, including but not limited to: **NumPy, pandas, matplotlib, statsmodels, and scikit-learn**. An explanation of what each of these libraries does and a detailed walkthrough of each library will be in the following chapters. To begin installation and setup, we need to know what Conda is.

**Conda** is an open-source package management and environment management system that runs on Windows, macOS, and Linux. Think of it as an app store combined with a virtual divider for your computer's coding workspaces. It was specifically created to solve the massive headache of installing complex data science tools. The two distinct but vital functions Conda serves:

* **Package manager**: downloads, installs, and updates software libraries (packages), containing C, C++, or R dependencies
* **Environment manager**: creates isolated folders (environments) on your computer for different projects. Imagine Enviro A --> Old project with Python 3.9, pandas 1.0; Enviro B --> New project with Python 3.13, pandas 2.5

## 1.2 Installing Necessary Packages
Now that we've set up Miniconda, it's time to install the main packages we'll be using. The first step is to configure **conda-forge** as your default package channel. 

Conda-forge is a community-led repository (or "channel") that provides open-source software packages and installers for the Conda package manager. If Conda is the app store on your computer, you can think of conda-forge as the **massive, community-driven catalog** where developers upload and maintain the absolute latest versions of data science and math tools.

Once this configuration is complete, any time you run a command like conda install pandas, your system will automatically search the conda-forge repository first to give you the most stable, up-to-date version available. Now, we'll create a new **conda environment** (note that "base" is your base environment) and activate the environment.

* Remember!! each time you open a new terminal, it is necessary to use **conda activate** to activate your environment

Now, we'll install the essential packages. There are two ways to install packages: with **conda install** and with **pip install**. conda install should always be preferred when using Miniconda, but some packages are not available through conda, in which case try pip install instead.

## 1.3 IPython, Jupyter, and Other Development Environments

### 1.3.1 IPython
IPython stands for **Interactive Python**. It is an upgraded, significantly more powerful version of the standard Python interactive terminal (the command-line window where you type Python code and get instant results). If the standard Python terminal feels like a basic digital notepad, IPython acts like a fully featured word processor designed specifically for data analysis, rapid prototyping, and experimentation.

IPython offers a series of advantages, including **syntax highlighting, tab completion, and detailed introspection (?)**. Besides, it also introduces unique, non-Python shortcut commands called **Magic Commands**. They always start with a percent sign (%) and let you control your computer or monitor your code easily.

### 1.3.2 Jupyter Notebook
Jupyter Notebook is a web-based, interactive computing environment that allows you to combine live code, computational math equations, rich explanatory text, and visual data charts into a single, cohesive document. The word Jupyter is a blend of Julia, Python, and R. The interactive code blocks inside a Jupyter Notebook run entirely on top of an **IPython background engine**.

A Jupyter Notebook saves your work as a file ending in the .ipynb extension (which stands for Interactive Python Notebook). Inside this file, you work with three primary types of building blocks (called cells):

* **Code cells**: Areas where you write true executable Python, R, or Julia code. When you run a code cell, its output (numbers, tables, or graphs) prints directly below the cell
* **Markdown cells**: Areas where you write plain explanatory text using simple formatting syntax. You can use bold text, lists, bullet points, and even beautifully typeset math equations using LaTeX notation
* **Raw cells**: Areas where you want to write text or code templates that you do not want to execute
