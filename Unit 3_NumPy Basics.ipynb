# Unit 3. NumPy Basics

NumPy (Numerical Python) is one of the core packages for numerical computing in Python. Pandas, Matplotlib, Statmodels and many other Scientific libraries rely on NumPy.

NumPy major contributions are:

* Efficient numeric computation with C primitives
* Efficient collections with vectorized operations
* An integrated and natural Linear Algebra API
* A C API for connecting NumPy with libraries written in C, C++, or FORTRAN.

Let's develop on efficiency. In Python, **everything is an object**, which means that even simple ints are also objects, with all the required machinery to make object work. We call them "Boxed Ints". In contrast, NumPy uses primitive numeric types (floats, ints) which makes storing and computation efficient.

## 3.1 Creating ndarrays
An ndarray is a generic multidimensional container for homogeneous data, where each array has a **shape** and **dtype**. With functions like np.zeros, you are able to create uni- or multi-dimenstional numerical arrays efficiently.

## 3.2 Data Types for ndarrays
Data types in ndarrays, like int 32 or int 64, is not the same as, but compatible with, the default data types in Python (as they provide mapping directly to underlying memory representation). You are able to cast between types with **.astype()**.

## 3.3 Basic Indexing and Slicing
Indexing of ndarray works similar to indexing of multidimensional lists in Python, and is more flexible.

## 3.4 Boolean Indexing
What's cool is that you can add conditional statements to indexing of ndarrays--think of it as applying a mask. So you can write something like **data[condition1 and condition2]**, which can be mixed with basic indexing.

## 3.5 ndarray Operations
Instead of writing excessive loops, NumPy allows you to perform quick element-wise or array-wise operations, a process often called **"broadcasting"**

### 3.5.1 Element-wise ndarray operations
Element-wise ndarray operations include simple arithmetics (+, *), unary universal-functions (ufuncs) like np.exp and np.isnan, binary ufuncs like np.add and np.fmax, and conditional logics like np.where.

### 3.5.2 Array-wise ndarray operations
Array-wise operations are useful in mathematical and statistical methods, and allows aggregation along a single axis (0 or 1). We'll explore a few commonly used ones:

* Statistical methods: np.mean(), np.sort(), np.unique() (beware of np.nan values)
* Boolean arrays: [condition].sum(), [condition].any(), [condition].all()

## 3.6 Pseurandom Number Generation
Random numbers are not truly random but instead are generated by a configurable random number generator in NumPy (which is typically faster than the Python random). Commonly used operations include:

* np.random.standard_normal(size)
* np.random.randint(low, high, size)
* np.random.default_rng(seed) (to configure a "fixed" rng)

## 3.7 Linear Algebra
NumPy supports linear algebra operations, like matrix multiplication, transposition, determinants, etc. Feel free to explore matrix operations on your own~
