**Core knowledge**: data types & structures, functions, files & operating system

**Your task**: Given a teacher's gradebook (a spreadsheet that contains the fields: student name, class, and grade), write the following functions. You may print out the results of your functions for testing.

* **getNames**: Get all student names and sort in alphabetical order, e.g. "Alice, Amy, Bob, Chris, ...".
* **getClasses**: Group the students by class names, e.g. "Class 1: Alice, Bob, ...".
* **sortByGrd**: Sort students by grade in descending order (from high to low), with students who didn't attend the exam placed at the end.
* **getRemakeLst**: Find all students who failed (grade < 60) or didn't attend the exam.
* sortByGrd_plus (optional): Sort students by grade and group them into intervals of 10 (100, 90 - 99, 80 - 89, etc.).

**Test gradebook**: You may use the following "grdBook1" or the example file "gradebook_ex.csv" for testing.

```python
grdBook1 = [
	{'Name':'Bob', 'Class':'Class 8', 'Grade':90},
	{'Name':'Cindy', 'Class':'Class 5', 'Grade':95},
	{'Name':'Aaron', 'Class':'Class 2', 'Grade':80},
	{'Name':'Rachael', 'Class':'Class 2', 'Grade':'N/A'},
	{'Name':'Henry', 'Class':'Class 8', 'Grade':55},
	{'Name':'Alice', 'Class':'Class 5', 'Grade':75},
	{'Name':'Eric', 'Class':'Class 3', 'Grade':60},
	{'Name':'Richard', 'Class':'Class 5', 'Grade':'N/A'}
]
path = "gradebook_ex.csv"
```

```python
# Your imports go here:
from itertools import groupby
import csv
```

```python
# Your function goes here:

def getNames(gradebook):
    names = sorted([std["Name"] for std in gradebook])
    print("**** Student Names ****")
    for name in names:
        print(f"   - {name}")
    return

def getClasses(gradebook):
    # Do sorting
    gradebook_sorted = sorted(gradebook, key=lambda x: x["Class"])
    print("**** Students by Class ****")
    for cls, stds in groupby(gradebook_sorted, key=lambda x: x["Class"]):
        print(f"     {cls}:")
        for std in stds:
            name = std["Name"]
            print(f"   - {name}")
    return 

def sortByGrd(gradebook):
    # Filter out students with N/A values
    std_NA = [std for std in gradebook if (std["Grade"] == "N/A")]
    std_eff = [std for std in gradebook if (std["Grade"] != "N/A")]
    std_sorted = sorted(std_eff, key=lambda x: x["Grade"], reverse=True)
    std_sorted.extend(std_NA)
    print("**** Student by Grade ****")
    for std in std_sorted:
        name = std["Name"]
        grade = std["Grade"]
        print(f"   - {name}" + " "*(10-len(name)) + f"{grade}")
    return std_sorted

def getRemakeLst(gradebook):
    # Filter out students with N/A values
    std_NA = [std for std in gradebook if (std["Grade"] == "N/A")]
    std_fail = [std for std in gradebook if (std["Grade"] != "N/A") and (std["Grade"] < 60)]
    std_fail.extend(std_NA)
    print("**** Remake List ****")
    for std in std_fail:
        name = std["Name"]
        grade = std["Grade"]
        print(f"   - {name}" + " "*(10-len(name)) + f"{grade}")
    return

def sortByGrd_plus(gradebook):
    gradebook_sorted = sortByGrd(gradebook)
    for std in gradebook:
        if std["Grade"] == "N/A":
            std["Category"] = "Absent"
        else:
            interval_floor = (std["Grade"] // 10) * 10
            interval_ceil = (std["Grade"] // 10) * 10 + 9
            std["Category"] = f"{interval_floor} - {interval_ceil}"
    print("**** Student by Grade Intervals ****")
    for interval, stds in groupby(gradebook_sorted, key=lambda x: x["Category"]):
        print(f"     {interval}:")
        for std in stds:
            name = std["Name"]
            grade = std["Grade"]
            print(f"   - {name}" + " "*(10-len(name)) + f"{grade}")
    return
```

```python
# Your test cases go here:
getNames(grdBook1)
getClasses(grdBook1)
sortByGrd(grdBook1)
getRemakeLst(grdBook1)
sortByGrd_plus(grdBook1)
```

```python
f = open(path, mode='r', encoding="utf-8")
def makeGradebook(f):
    rows = csv.DictReader(f)
    gradebook = []
    for row in rows:
        std = {}
        std["Name"] = row["Name"]
        std["Class"] = row["Class"]
        try:
            std["Grade"] = int(float(row["Grade"]))
        except:
            std["Grade"] = "N/A"
        gradebook.append(std)
    return gradebook

grdBook2 = makeGradebook(f)
getNames(grdBook2)
getClasses(grdBook2)
sortByGrd(grdBook2)
getRemakeLst(grdBook2)
sortByGrd_plus(grdBook2)
```
