**Your task**: Write a function isPrime that takes in a positive integer and determines whether the integer is a prime.

* If the input is valid (positive integer), the function should return True if the integer is prime and False otherwise.
* For invalid inputs (such as wrong data types, negative integers), the function should return a warning like "Invalid input: must be positive integer" instead of raising an error.
* You should write test cases for your function, including true, false, invalid, and edge cases.

**Optional task**: Improve the mathematical algorithm of your function, for instance, by applying the Sieve of Eratosthenes. 

* You may test whether it has been "improved" by calling the %timeit function. 

```python
# Your imports go here:
import math
```

```python
# Your function goes here:

def isPrime(num):
    if not(isinstance(num, int)) or (num <= 0): # Check for invalid inputs
        return "Invalid input: must be positive integer"
    if (num == 1):
        return False
    for i in range(2, num):
        # Multiple of a factor between 2 and num-1
        if (num % i == 0):
            return False
    return True

def isPrime_SoE1(num):
    if not(isinstance(num, int)) or (num <= 0): # Check for invalid inputs
        return "Invalid input: must be positive integer"
    if (num == 1):
        return False
    sqrt_num = math.floor(math.sqrt(num))
    for i in range(2, sqrt_num + 1):
        # Multiple of a factor between 2 and sqrt of num
        if (num % i == 0):
            return False
    return True
```

```python
def getPrimes(num):
    # A list of prime numbers not exceeding num
    lst = []
    for i in range(2, num+1):
        if isPrime(i):
            lst.append(i)
    return lst

getPrimes(100)

def getPrimes_SoE(num):
    if (num == 1):
        return set()
    sqrt_num = math.floor(math.sqrt(num))
    prime_set = getPrimes_SoE(sqrt_num)
    remain_set = set(range(sqrt_num + 1, num + 1))
    remove_set = set()
    for prime in prime_set:
        start = math.ceil((sqrt_num + 1) // prime) * prime
        for prime_multiple in range(start, num + 1, prime):
            remove_set.add(prime_multiple)
    prime_set.update(remain_set - remove_set)
    print(f"n = {num}: {prime_set}")
    return prime_set

getPrimes_SoE(100)

def getPrimes_2(num): # Takes in a positive integer num and returns all primes not exceeding num
    if (num == 1): 
        return []
    else:
        sqrt_num = math.floor(math.sqrt(num))
        sqrt_lst = getPrimes_2(sqrt_num) # all primes not exceeding square root of n
        mask = [1] * (num - sqrt_num)
        for prime in sqrt_lst:
            start = (math.ceil((sqrt_num + 1)/prime)) * prime
            for prime_multiple in range(start, num + 1, prime):
                mask[prime_multiple - sqrt_num - 1] = 0
        for i in range(num - sqrt_num):
            if mask[i] == 1:
                sqrt_lst.append(i + sqrt_num + 1)
        # print(f"n = {num}: {sqrt_lst}")
        return sqrt_lst

%timeit getPrimes_2(10000000)
```
