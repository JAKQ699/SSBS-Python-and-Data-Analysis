## 5.1 Data Loading and Writing
Input and output typically fall into a few main categories: reading text files and other more efficient on-disk formats, loading data from databases, and interacting with network sources like web APIs. We'll focus on csv files as they are most commonly used in data analysis.

### 5.1.1 Reading text files
Remember we've discussed the Python's default open(path). **pd.read_csv** is much more convenient, while there are some typical arguments to be concerned about:

* Indexing: columns/rows that need to be read, including **names**, **index_col**, and **skiprows**
* NA handling: some initial null value handling like **na_values** (you may specify null values as "sentinels"), **keep_default_na**
* Type inference and conversion: specify data type with **dtype**
* Date and time parsing: automatically parse time and dates with **parse_dates**, discussed later in time series
* Iterating: reading file in pieces, like specified number of rows (**nrows**) or iterable chunks (**chunksize**)
* Other arguments like **delimiter** and **encoding**

### 5.1.2 Writing data to text format
Data can also be exported to a delimited format (though not typical) with **data.to_csv()**. You may consider displaying the result in system console with **data.to_csv(sys.stdout)** instead.

### 5.1.3 Other delimited formats
We'll discuss JSON and HTML/XML.

#### JSON
JSON (short for JavaScript Object Notation) has become one of the standard formats for sending data by HTTP request between web browsers and other applications. It is nearly valid Python code with the exception of its null value null and some other nuances.

Binary data formats store information as sequence-of-bytes data (0s and 1s) rather than plain text characters. They are optimized for machines to read and write quickly, making them significantly faster and more compact than text formats like JSON or XML.

## 5.3 Extension Data Types
Remember pandas was originally built upon the capabilities present to NumPy, but that also led to a number of shortcomings, such as:

* Missing data handling for some numerical data types, like integers and booleans, was incomplete
* Datasets with a lot of string data were computationally expensive
* Some data types, like timestamps and time intervals, could not be supported efficiently. This will be explored further when we talk about time series

Therefore, pandas has developed an **extension type** system to support new data types alongside data coming from NumPy. Some commonly used ones include "Int64", "string", "boolean", "Categorical" (explored later).

### 5.3.1 Categorical extension type
pandas has a special **Categorical** extension type for holding data that uses the **integer-based categorical representation** or encoding. It can provide significantly faster performance with lower memory use, especially for string data.

Categorical data can be created directly from string Series or Dataframe, and have attributes **categories** and **codes**.

### 5.3.2 Creating dummy variables
When using statistics or machine learning tools, it is a common practice to transform categorical data into **dummy variables**, also known as one-hot encoding (1s for occurrence of a given category and 0 otherwise). This can be achieved through **pd.get_dummies()**.

## 5.4 Advanced String Manipulation
Handling (long and complex) strings is often among the first steps of data preparation. We've already encountered some built-in Python string methods like str.strip or str.replace.

### 5.4.1 Regular expressions
Regular expressions provide a flexible way to search or match (often more complex) string patterns in text. Python's built-in **re** module is responsible for applying regular expressions to strings.

When using the re module, the first thing to do is to compile re with a desired pattern with: **regex = re.compile(pattern)**, and then the compiled regex can act on the target string with methods like regex.findall(str), regex.search(str), regex.match(str).

### 5.4.2 String functions in pandas
pandas Series has built-in **str** attribute (it's recommended to convert to extension type "string" first), which contains methods similar to Python built-ins or regular expressions.
