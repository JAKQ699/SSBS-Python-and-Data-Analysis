```python
# Your test cases go here:
data1 = pd.read_csv(path1, na_values=-1) # sentinel values for NA
data2 = pd.read_csv(path2, index_col = ["Name", "Class"])
# data1.replace(-1, pd.NA)
data1
data3 = pd.read_csv(path1, na_values=-1, index_col = ["Name", "Class"])
data3
```

```python
# Find if there's any repetitive names
name_count = data1["Name"].value_counts()
name_count[name_count != 1]
```

```python
# Find the number of students in each class
print(data1["Class"].value_counts().sort_index())
for key, group in data1.groupby("Class")["Name"]: # Aggregate function
    print(f"{key}:")
    print(f"{list(group)}")
```

```python
exam_anal = data1.describe().round(2)
# count: the count of non-NA values (students who actually attend)
exam_anal.loc["mean"]
# sort the 5 exams by their average grade in descending order
exam_anal.sort_values("mean", axis="columns", ascending=False)
```

```python
# Student sorted by grade in descending order
data3.T
std_anal = data3.T.describe().round(2)
std_anal.sort_values("mean", axis="columns", ascending=False).T
```

```python
grades = data1.loc[:, "Grade 1":"Grade 5"]
# Whether each student has at least a fail
fail = grades[grades < 60].any(axis="columns")
# Whether each student has at least an absence
absence = grades.isna().any(axis="columns")
print(fail | absence)

fails = grades[grades < 60].count(axis="columns") # count the number of fails for each student
absences = grades.isna().sum(axis="columns") # count the number of absences for each student
print(fails + absences)

# Make-up list: at least two fails or absences
makeup = fails + absences
makeup[makeup >= 2]
data1["Name"][makeup[makeup >= 2].index]
```
