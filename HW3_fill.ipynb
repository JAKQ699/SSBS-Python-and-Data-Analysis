**Your task**: This time, you will be given a teacher's gradebook that contains student names, classes, and their grades of 5 exams (with some potential absences represented as "-1"). You are asked to perform a comprehensive exam analysis as well as student analysis. 

* **Exam analysis**: including but not limited to calculating the average grade, max grade, min grade, standard deviation, and number of absences for each of the 5 exams. All float values should be rounded to 2 decimal places. 
* **Student analysis**: including but not limited to calculating the average grade across the 5 exams, sorting the students by average grade, and counting the number of fails (grade < 60) or absences to determine whether the student need to make up (at least 2 fails or absences).

**Test gradebook**: You may use the example file "gradebook_ex2.csv" for testing.

```python
path = "gradebook_ex2.csv"
```

```python
# Your imports go here:
import numpy as np
import pandas as pd
```

```python
# Your functions go here:
def examAnalysis(gradebook, exams):
    avg = np.round(np.nanmean(gradebook, axis=0), 2)
    mx = np.nanmax(gradebook, axis=0)
    # mx_id = np.nanargmax(gradebook, axis=0) student who scores the highest
    mn = np.nanmin(gradebook, axis=0)
    std = np.round(np.nanstd(gradebook, axis=0), 2)
    num_abs = np.isnan(gradebook).sum(axis=0)
    print("**** Exam Analysis ****")
    for i in range(len(exams)):
        print("\n")
        print(f"     {exams[i]}")
        print(f"   - Average: {avg[i]}")
        print(f"   - Max: {mx[i]}")
        print(f"   - Min: {mn[i]}")
        print(f"   - Standard Deviation: {std[i]}")
        print(f"   - Number of Absences: {num_abs[i]}")
    return

def studentAnalysis(gradebook, students):
    avg = np.round(np.nanmean(gradebook, axis=1), 2)
    id_sort = np.argsort(avg)[::-1]
    num_abs = np.isnan(gradebook).sum(axis=1)
    num_fails = (gradebook < 60).sum(axis=1)
    num = num_abs + num_fails # total number of absences or fails
    remake = np.where((num >= 2), "Yes", "No")
    print("**** Student Analysis ****")
    for i in id_sort:
        print("\n")
        print(f"     {students[i]}")
        print(f"   - Average: {avg[i]}")
        print(f"   - Number of Absences or Fails: {num[i]}")
        print(f"   - Need to Remake?: {remake[i]}")
    return
```

```python
# Your test cases go here:
gradebook_raw = pd.read_csv(path, na_values=-1) # na_value: sentinel values for missing values
print(gradebook_raw)
exm = ["Grade 1", "Grade 2", "Grade 3", "Grade 4", "Grade 5"]
exm_names = ["Exam 1", "Exam 2", "Exam 3", "Exam 4", "Exam 5"]
std_names = gradebook_raw["Name"]
grdBook = gradebook_raw[exm].to_numpy()
grdBook
```

```python
examAnalysis(grdBook, exm_names)
print("\n")
studentAnalysis(grdBook, std_names)
```
